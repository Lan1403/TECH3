```python
from scipy import stats
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,1,2000)
ppts = np.linspace(0,1,2000)
pdf = stats.uniform.pdf(x)
cdf = stats.uniform.cdf(x)
ppf = stats.uniform.ppf(ppts)

plt.plot(x,pdf)
plt.ylabel("Uniform density")
plt.xlabel("x")
plt.show()
```

```python
plt.plot(x,cdf)
plt.xlabel('x')
plt.ylabel("Uniform cumulative distribution function")
plt.show()
```

```python
plt.plot(100*ppts, ppf)
plt.ylabel("Uniform quantile function")
plt.ylabel("Percentage point (%)")
plt.show()
```

```python
x = np.linspace(0,1,2000)
ppts = np.linspace(-5,5,2000)
pdf = stats.norm.pdf(x)
cdf = stats.norm.cdf(x)
ppf = stats.norm.ppf(ppts)

plt.plot(x,pdf)
plt.title("Normal density")
plt.show()
```

```python
plt.plot(x,cdf)
plt.title("Normal cumulative distribution function")
plt.show()
```

```python
plt.plot(100*ppts, ppf)
plt.title("Normal quantile function")
plt.show()
```

Resampling and bootstrap

```python
import pandas as pd
numbers = [183, 179, 179,183,178,176,182,177,180,179]
vector = pd.Series(numbers)
newsample = vector.sample(3, replace = True)
print (newsample)
```

```python
print(vector.mean())
```

```python
import math
print(vector.std()/math.sqrt(10))
```

```python
np.random.seed(123)
bootstrap_samples = 2000
boot_means = np.zeros(bootstrap_samples)
for b in range (0, bootstrap_samples):
    boot_sample = vector.sample(len(vector), replace = True)
    boot_means[b] = boot_sample.mean()
print(boot_means.std())
```

```python
print(boot_means)
```

```python
from scipy import stats
xvalues = np.linspace(177, 182, 2000)
pdf = stats.norm.pdf(xvalues, loc= boot_means.mean(), scale = boot_means.std())
sns.histplot(boot_means, stat = "density", binwidth= 0.2)
plt.axvline(x=vector.mean(), color= "red")
plt.show(xvalues, pdf, color = "green")
plt.show()
```
