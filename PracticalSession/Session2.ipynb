Problem 1

Let X be the number of dots on the dice in the experiment of our respondent. Waht are the possible values of X and what is the probability distribution function of X?

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
import numpy as np
```

```python
column_data = data['reported_diceroll']
frequence_distribution = column_data.value_counts(normalize = True)
sns.barplot(x = frequence_distribution.index, y = frequence_distribution.values, palette = 'viridis')
plt.xlabel('Values')
plt.ylabel('Probability')
plt.title('Probability Distribution')
plt.show()
```

The respondents have no incentive to lie about the outcome of the dice. Find the relative frequency of the various reported outcomes of the dice and make a visualization of the distribution. Is it consistent with your answer in Problem 1.

```python
correct = data[data['correct_guess'] == 1 ]

reported_diceroll = correct ['reported_diceroll']
frequency = reported_diceroll.value_counts()
sns.barplot(x=frequency.index, y= frequency.values, palette = 'rocket')

plt.xlabel('values on dice')
plt.ylabel('frequency')
plt.title ('frequence distribution')
plt.show()
```
